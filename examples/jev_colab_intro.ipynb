# Jev Evidence Kit: count avoided tokens only after the answer passes

This short exercise uses **synthetic examples only**. It demonstrates one narrow rule: credit avoided ChatGPT tokens only when both the baseline and local result exactly match the expected answer. It does not assess semantic quality, verify a provider bill, or benchmark a model.

Colab runs code on Google's hosted runtime. Keep private prompts, customer data, and secrets out of this notebook. The downloadable Jev CLI runs locally instead.

## 1. Define two synthetic pairs

The first pair passes. The second uses fewer tokens but the local answer is wrong, so it receives zero credit.

```python
pairs = [
    {
        'name': 'passing example',
        'expected': 'The request succeeded.',
        'baseline': 'The request succeeded.',
        'local': 'The request succeeded.',
        'baseline_chatgpt_tokens': 150,
        'local_chatgpt_tokens': 0,
    },
    {
        'name': 'cheaper but incorrect',
        'expected': 'The request succeeded.',
        'baseline': 'The request succeeded.',
        'local': 'The request failed.',
        'baseline_chatgpt_tokens': 150,
        'local_chatgpt_tokens': 0,
    },
]
```

```python
def normalize(text):
    return text.replace('\r\n', '\n').strip()

def evaluate(pair):
    baseline_pass = normalize(pair['baseline']) == normalize(pair['expected'])
    local_pass = normalize(pair['local']) == normalize(pair['expected'])
    avoided = max(0, pair['baseline_chatgpt_tokens'] - pair['local_chatgpt_tokens']) if baseline_pass and local_pass else 0
    return {
        'name': pair['name'],
        'baseline_pass': baseline_pass,
        'local_pass': local_pass,
        'chatgpt_tokens_avoided_credited': avoided,
    }

for pair in pairs:
    print(evaluate(pair))
```
